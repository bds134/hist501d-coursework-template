## Quick Lessons: Coordinate Systems

### Geographic Coordinate Systems (GCS)
- **GCS** use latitude and longitude to locate points on the curved surface of the Earth.
- Locations are measured in angular units from the Earth’s center, relative to the equator and prime meridian.

![Angular Units in GCS](https://mgimond.github.io/Spatial/img/Colby_location_sphere.jpg)

**Latitude and Longitude:**

Different notations exist for latitude and longitude, but they all represent the same concept of angular measurements on the Earth's surface.

Here we have angular measurements:

![Latitudinal and longitudinal lines](https://mgimond.github.io/Spatial/img/lat_vs_lon.png)

For web mapping, we often use decimal degrees with positive values for north/east and negative for south/west.

NYC LATITUDE: 40.714 LONGITUDE: -74.006 ![Decimal degrees](https://gisgeography.com/wp-content/uploads/2017/11/DecimalDegrees-NewYork.png)

**Ellipsoid vs Sphere:**
- The Earth is not a perfect sphere; it is slightly flattened at the poles (an ellipsoid).
- The semi-major axis (equator) is about 21 km longer than the semi-minor axis (pole).

![Ellipsoid axes](https://mgimond.github.io/Spatial/img/Ellipsoid.svg)
![Sphere vs Ellipsoid](https://mgimond.github.io/Spatial/img/globe_0_0_center_wide.png)

**Geoid:**
- The geoid models the Earth's true shape, accounting for gravitational variations.

![Earth's geoid](https://mgimond.github.io/Spatial/img/geoid.png)

**Datum:**
- A datum aligns the ellipsoid with the geoid, either locally or geocentrically.
- Examples: NAD27 (local), NAD83 and WGS84 (geocentric).

![Datum alignment](https://mgimond.github.io/Spatial/img/datum.png)

**NB:** Web mapping typically uses WGS84, which is the standard for GPS and online maps. **However**, if you get data from a national or local source, it may be in a different datum, so you may need to reproject it to WGS84 for accurate mapping.

### Vector vs Raster Data Models

> Adapted from [Spatial: Feature Representation](https://mgimond.github.io/Spatial/chp02_0.html#vector-vs.-raster) by M. Gimond

**Vector Model:**
- Represents discrete features as points, lines (polylines), and polygons.
- Each feature is defined by coordinates (X, Y) and can have rich attribute data.
- Common for roads, buildings, boundaries.
- Examples:
    - **Point:**
  
      ![Three point objects](https://mgimond.github.io/Spatial/02-Feature-representation_files/figure-html/unnamed-chunk-1-1.png)

    - **Polyline:**

      ![Simple polyline](https://mgimond.github.io/Spatial/02-Feature-representation_files/figure-html/unnamed-chunk-2-1.png)

    - **Polygon:**

      ![Simple polygon](https://mgimond.github.io/Spatial/02-Feature-representation_files/figure-html/unnamed-chunk-3-1.png)

**Raster Model:**
- Represents continuous data as a grid of cells (pixels), each with a value.
- Used for imagery, elevation, temperature, land cover.
- Each cell covers a specific area and stores a measurement or classification.
- Example:

![Simple raster object](https://mgimond.github.io/Spatial/02-Feature-representation_files/figure-html/unnamed-chunk-4-1.png)

**Comparison:**
- Vector is best for precise, discrete features; raster is best for continuous surfaces.
- The same feature (e.g., a river) can be represented in both ways:

![Vector and raster representations of a river](https://mgimond.github.io/Spatial/img/vector_vs_raster.jpg)

### 1) GeoJSON
**GeoJSON** is a JSON-based format for vector geospatial data.  
It stores geometries such as:
- `Point`
- `LineString`
- `Polygon`
- collections of features with attributes

**Why it is useful:**
- Human-readable and web-friendly
- Works well with JavaScript tools (Leaflet, Mapbox)
- Common for APIs and lightweight map layers

### 2) TEI XML (Text Encoding Initiative)
**TEI XML** is primarily a scholarly text-encoding format, not a native GIS format.  
However, it is useful in digital humanities because it can encode:
- place names (`<placeName>`)
- references to gazetteers
- optional coordinates and spatial metadata

**Why it is useful:**
- Excellent for historical documents where geography is embedded in text
- Supports rich semantic markup and provenance
- Can be transformed into GIS-ready data for mapping workflows
- Can be used to link textual descriptions to spatial features in a GIS

#### XML

- Hierarchical data structure that can represent complex relationships
- Not designed for spatial data but can include spatial metadata

**Basic non-spatial XML Example**

```xml
<note>
  <to>Alice</to>
  <from>Bob</from>
  <heading>Reminder</heading>
  <body>Don't forget our meeting tomorrow!</body>
</note>
```

#### TEI

[Text Encoding Initiative](https://www.tei-c.org/)

>The TEI Consortium is a nonprofit membership organization composed of academic institutions, research projects, and individual scholars from around the world. We develop the Guidelines, which provide the infrastructure for developing machine-actionable cultural heritage texts. Since 1994, the TEI Guidelines have been widely used by libraries, museums, publishers, and individual scholars to present texts for online research, teaching, and preservation.

TEI uses tags from a progenitor of HTML called SGML (Standard Generalized Markup Language), which was developed in the 1980s for encoding documents in a structured way. TEI adapted and extended SGML to create a rich set of tags for encoding literary and historical texts, including spatial information. There are hundreds of TEI tags available; HTML uses a smaller subset of these tags.

TEI tags are **strictly semantic** and are designed to encode the meaning and structure of texts, rather than their visual presentation. This allows TEI-encoded texts to be transformed and displayed in various ways while preserving their underlying semantic information.

**Standard Tags used in TEI**

```xml
<?xml version="1.0" encoding="UTF-8"?>
<TEI xmlns="http://www.tei-c.org/ns/1.0">
  <teiHeader>
    <fileDesc>
      <titleStmt>
        <title>A Sample TEI Document</title>
        <author>Jane Doe</author>
      </titleStmt>
      <publicationStmt>
        <p>Electronic edition for demonstration purposes.</p>
      </publicationStmt>
      <sourceDesc>
        <p>Originally written as an example for a guide.</p>
      </sourceDesc>
    </fileDesc>
  </teiHeader>
  <text>
    <body>
      <div type="chapter">
        <head>Chapter 1: The Beginning</head>
        <p>This is a standard paragraph in a TEI document. 
           It can include semantic tags like <persName>Jane Doe</persName> 
           to identify people or <date when="2024-03-20">March 20th</date> 
           for specific dates.</p>
        <p>You can also mark graphical features, such as 
           <hi rend="italic">italics</hi> or <emph>emphasized text</emph>.</p>
      </div>
    </body>
  </text>
</TEI>
```

**Example TEI XML with Geospatial Markup:**

### 3) Shapefiles
A **Shapefile** is a classic vector GIS format (actually a set of files, e.g., `.shp`, `.shx`, `.dbf`, `.prj`), typically downloaded as a ZIP archive.

**Why it is useful:**
- Widely supported across GIS software
- Good for storing points, lines, and polygons with attributes
- Common in government and institutional data portals

**Limitations:**
- Older format with naming/size constraints
- Requires multiple companion files
- Not easily human-readable or web-friendly compared to formats like GeoJSON

---

### 4) GeoTIFF
A **GeoTIFF** is a raster image format (TIFF) with embedded georeferencing metadata (CRS, extent, pixel size).

**Why it is especially useful for georectified historical maps:**
- Scanned historical maps can be georeferenced to real-world coordinates
- Preserves the original cartographic image while enabling spatial analysis
- Can be overlaid with modern vector/raster layers in GIS tools
- Supports archival-quality raster data and metadata

In short, GeoTIFF is the standard format when working with **georectified historical map scans**.

Good sources for georectified historical maps include:
- [David Rumsey Map Collection](https://www.davidrumsey.com/)
- [Library of Congress Map Collections](https://www.loc.gov/maps/)
- [USGS Historical Topographic Map Explorer](https://livingatlas.arcgis.com/topoview/)
- NYPL Map Warper (https://maps.nypl.org/warper/) for georectifying your own maps

Georectification is the process of aligning a scanned map to real-world coordinates, allowing it to be used in GIS applications alongside other spatial data.

Good tools for georectification include:
- QGIS Georeferencer plugin
- ArcGIS Georeferencing tools
- Map Warper (web-based)


## Leaflet.js: A JavaScript Library for Interactive Maps

**Leaflet.js** is a popular open-source JavaScript library for creating interactive maps on the web. It is lightweight, easy to use, and supports a wide range of features for displaying geospatial data.

[Leaflet.js Documentation](https://leafletjs.com/)

### A Few Words about JavaScript

JavaScript is a programming language that runs in web browsers and allows developers to create interactive and dynamic web applications. It is one of the core technologies of the web, alongside HTML and CSS.

JavaScript relies upon a runtime environment (like a web browser) to execute code. This means that JavaScript code is typically executed on the client side (in the user's browser), allowing for dynamic interactions without needing to reload the page.

Do not confuse JavaScript with Java, which is a different programming language. JavaScript was originally developed by Netscape in the mid-1990s and was initially called "LiveScript" before being renamed to "JavaScript" for marketing reasons. Despite the name, JavaScript has no direct relation to Java.

In an HTML document, you can include JavaScript code within `<script>` tags, which can be placed in the `<head>` or `<body>` sections. These scripts will be read

### Steps to Create a Basic Leaflet Map
1. **Include Leaflet CSS and JS** in your HTML file:

```html
<head>
  <!-- Leaflet’s stylesheet should go in the head so the map tiles are styled as soon as they start downloading. -->
  <link rel="stylesheet" href="https://unpkg.com/leaflet/dist/leaflet.css" />

  <!-- The Leaflet script can also live in the head, but for best performance people often move `<script>` tags to just before `</body>`. -->
  <script src="https://unpkg.com/leaflet/dist/leaflet.js"></script>
</head>
```

2. **Create a Map Container** in your HTML:

```html
<div id="map" style="height: 500px;"></div>
```

This `div` will be the container for your map, and you can set its height and width using CSS. The `div` element must be placed in the appropriate location in your HTML body where you want the map to appear.

3. **Initialize the Map** in your JavaScript:

```javascript
var map = L.map('map').setView([40.7128, -74.0060], 13); // Centered on NYC
L.tileLayer('https://{s}.tile.openstreetmap.org/{z}/{x}/{y}.png', {
    attribution: '&copy; OpenStreetMap contributors'
}).addTo(map);
```

4. **Add Markers or GeoJSON Layers**:

```javascript
// Adding a marker
L.marker([40.7484, -73.9857]).addTo(map)
    .bindPopup('Empire State Building')
    .openPopup();
// Adding a GeoJSON layer
var geojsonFeature = {
    "type": "Feature",
    "geometry": {
        "type": "Point",
        "coordinates": [-73.9857, 40.7484]
    },
    "properties": {
        "name": "Empire State Building"
    }
};
L.geoJSON(geojsonFeature).addTo(map);
```

````python
# Make basic folium map centered on Empire State Building
# Add a marker for the Empire State Building

import folium

map = folium.Map(location=[40.7488, -73.9857], zoom_start=13)
folium.Marker([40.7488, -73.9857], popup="Empire State Building").add_to(map)

# save the map to an HTML file
map.save("empire_state_map.html")
````

````python
# Add all features from demo-geojson.geojson to the folium map
import folium
import json

# Load GeoJSON data from file
with open('demo-geojson.geojson', 'r', encoding='utf-8') as f:
    geojson_data = json.load(f)

# Function to create popups from feature properties
def popup_function(feature):
    return folium.Popup(feature['properties']['name'], parse_html=True)

# Add GeoJSON layer with popups for each feature
folium.GeoJson(
    geojson_data,
    name="NYC Landmarks",
    popup=folium.GeoJsonPopup(fields=["name", "category"], labels=True)
).add_to(map)

# Save the map
map.save("empire_state_map_with_landmarks.html")
````

### Matplotlib
Matplotlib can be used to create static maps by plotting spatial data on a coordinate grid. You can use it in conjunction with Geopandas to visualize geospatial data.

````python
# Visualize NYC landmarks from demo-geojson.geojson using geopandas and matplotlib
import geopandas as gpd
import matplotlib.pyplot as plt

# Load GeoJSON data into a GeoDataFrame
gdf = gpd.read_file('demo-geojson.geojson')

# Plot the points
fig, ax = plt.subplots(figsize=(8, 8))
gdf.plot(ax=ax, color='dodgerblue', markersize=80, edgecolor='black')

# Annotate each point with its name
for idx, row in gdf.iterrows():
    ax.annotate(row['name'], xy=(row.geometry.x, row.geometry.y), xytext=(3, 3), textcoords='offset points', fontsize=9)

ax.set_title('NYC Landmarks and Points (GeoJSON)', fontsize=14)
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
plt.grid(True)
plt.show()
````

````python
# Now let's add a basemap using contextily
import contextily as ctx
# Convert GeoDataFrame to Web Mercator for contextily
gdf_web_mercator = gdf.to_crs(epsg=3857)
fig, ax = plt.subplots(figsize=(8, 8))
gdf_web_mercator.plot(ax=ax, color='dodgerblue', markersize=80, edgecolor='black')
# use an available provider, e.g. OpenStreetMap
ctx.add_basemap(ax, source=ctx.providers.OpenStreetMap.Mapnik)

plt.show()
````

### Geopandas

### Converting from TEI XML to CSV using Geopandas (via parsing XML)

One last conversion might be to take a TEI XML file with place names and coordinates and convert it to a CSV file. These conversions are essential for data interoperability. We can use the `lxml` library to parse the XML and extract the relevant information, then use `pandas` to create a DataFrame and save it as a CSV.

````python
# Convert TEI XML placenames to CSV using geopandas (via parsing xml)
import pandas as pd
# lxml is a powerful library for parsing XML and HTML in Python; it is included in the standard python library, but you may need to install it separately using pip if it's not already available in your environment.
from lxml import etree

# parse the TEI file
tree = etree.parse('demo-nyc-guide-tei.xml')

# find all placeName elements
places = tree.findall('.//{http://www.tei-c.org/ns/1.0}placeName')

records = []
for p in places:
    name = p.text
    lat = p.get('lat')
    lon = p.get('lon')
    ref = p.get('ref')
    records.append({'name': name, 'ref': ref, 'lat': float(lat), 'lon': float(lon)})

tei_df = pd.DataFrame(records)
tei_df.to_csv('tei_placenames.csv', index=False)
tei_df
````

## Geocoding and Reverse Geocoding

Geocoding is the process of converting addresses or place names into geographic coordinates (latitude and longitude). Reverse geocoding is the opposite process, where you convert geographic coordinates back into a human-readable address or place name.

### Geocoding with Python
You can use the `geopy` library to perform geocoding and reverse geocoding in Python. `geopy` provides a simple interface to various geocoding services, such as Nominatim (OpenStreetMap), Google Geocoding API, and more.

Below we will use the Nominatim geocoder, which is free and does not require an API key, but be mindful of usage limits. It interacts with OpenStreetMap, an ambitious project to create a free editable map of the world. Anyone can contribute to OpenStreetMap by adding or editing map data, making it a valuable resource for geocoding and mapping applications. Think of it as a collaborative mapping project where volunteers contribute geographic data, which can then be accessed and used through APIs like Nominatim for geocoding purposes.

Here's how to do it:

```bash
pip install geopy
```


````python
# Let's geocode the place names from the TEI XML file using geopy and save results to CSV
# We'll keep the original XML lat/lon and compare with geocoded values

from geopy.geocoders import Nominatim

geolocator = Nominatim(user_agent="digital_history_geocoder")
geocoded_records = []
for idx, row in tei_df.iterrows():
    name = row['name']
    orig_lat = row['lat']
    orig_lon = row['lon']
    try:
        location = geolocator.geocode(name + ", New York City")
        if location:
            geocoded_records.append({
                'name': name,
                'orig_latitude': orig_lat,
                'orig_longitude': orig_lon,
                'geo_latitude': location.latitude,
                'geo_longitude': location.longitude,
                'address': location.address
            })
        else:
            geocoded_records.append({
                'name': name,
                'orig_latitude': orig_lat,
                'orig_longitude': orig_lon,
                'geo_latitude': None,
                'geo_longitude': None,
                'address': None
            })
    except Exception as e:
        print(f"Error geocoding {name}: {e}")
        geocoded_records.append({
            'name': name,
            'orig_latitude': orig_lat,
            'orig_longitude': orig_lon,
            'geo_latitude': None,
            'geo_longitude': None,
            'address': None
        })

geocoded_df = pd.DataFrame(geocoded_records)
geocoded_df.to_csv('geocoded_tei_placenames.csv', index=False)
geocoded_df
````

Now imagine that we have a list of lat/lon coordinates and we want to find out what places they correspond to. We can use reverse geocoding for this purpose.

````python
# Reverse geocode the original lat/lon coordinates to get place names and compare with TEI names
reverse_geocoded_records = []
for idx, row in tei_df.iterrows():
    name = row['name']
    orig_lat = row['lat']
    orig_lon = row['lon']
    try:
        location = geolocator.reverse((orig_lat, orig_lon), exactly_one=True)
        if location:
            reverse_geocoded_records.append({
                'name': name,
                'orig_latitude': orig_lat,
                'orig_longitude': orig_lon,
                'rev_geo_name': location.address
            })
        else:
            reverse_geocoded_records.append({
                'name': name,
                'orig_latitude': orig_lat,
                'orig_longitude': orig_lon,
                'rev_geo_name': None
            })
    except Exception as e:
        print(f"Error reverse geocoding {name}: {e}")
        reverse_geocoded_records.append({
            'name': name,
            'orig_latitude': orig_lat,
            'orig_longitude': orig_lon,
            'rev_geo_name': None
        })
reverse_geocoded_df = pd.DataFrame(reverse_geocoded_records)
reverse_geocoded_df.to_csv('reverse_geocoded_tei_placenames.csv', index=False)
reverse_geocoded_df
````
