## Step 1: Design Your Database

The first part of the exercise, you will create a database of your own design. I recommend that you think about a topic that interests you and relates to your research. Indeed, you might already have a flat spreadsheet of data that you can split into multiple tables.

How do you know how to split your data into multiple tables? The best way to do this is to think about the entities in your data and the relationships between those entities. Let's think about a real example of a historical database. Let's say you are interested in the history of the American Civil War and you have a spreadsheet of data about the letters that soldiers wrote to their families during the war and about the letters that soldiers received from their families during the war. Soldiers moved from place to place during the war, so you might want to have a table of soldiers, a table of places, and a table of letters. The soldiers table would have information about each soldier, such as their name, rank, and regiment. The places table would have information about each place, such as its name and location. The letters table would have information about each letter, such as the date it was written, the sender, the recipient, and the content of the letter. The letters table would also have foreign keys that link it to the soldiers table and the places table. There are also individuals who wrote letters to soldiers, but they are not soldiers themselves, so you might want to have a separate table for those individuals and a join table that links them to the letters they wrote.

In plain terms, without any SQL script, here is a possible schema for this database:

SOLDIERS
- id (primary key)
- name (varchar)
- rank (varchar)
- regiment (varchar)

PLACES
- id (primary key)
- name (varchar)
- location (varchar)
- latitude (float)
- longitude (float)
- location description (varchar)

LETTERS
- id (primary key)
- date_written (date)
- sender_id (foreign key to soldiers)
- recipient_id (foreign key to soldiers)
- content (transcribed text)
- sent_from (foreign key to places)
- sent_to (foreign key to places)

INDIVIDUALS
- id (primary key)
- name (varchar)
- notes (varchar)

RELATIONSHIP TABLE
- id (primary key)
- relationship_type (varchar)

There are many different types of relationships that exist between these tables that might have one to many (1:M) or many to many relationships (M:N). For each of the latter types, each such relationship would require a join table to model it. For a 1:M relationship, you can just add a foreign key to the table on the "many" side of the relationship. 

A soldier goes to many **places** during the war, and a place has many soldiers who were there, so that is a M:N relationship that would require a join table to model it.

SOLDIER_PLACES
- soldier_id (foreign key to soldiers)
- place_id (foreign key to places)

A soldier writes many **letters**, but the letter is only written by one soldier, so that is a 1:M relationship that can be modeled with a foreign key in the letters table that references the soldiers table. The same goes for the recipient of the letter, which is just one person, so that is also a 1:M relationship that can be modeled with a foreign key in the letters table that references the soldiers table.

And then there are the relationships between letter writers. It might be interesting to track these relationships. For example, a soldier might have a mother, a father, a spouse, and a sibling who all wrote letters to the soldier during the war. The soldier might also have friends who wrote letters to the soldier during the war. These relationships are all M:N relationships that would require a join table to model them.

In this case, we would have a join table with an attribute that describes the type of relationship (e.g. mother, father, spouse, sibling, friend, etc.) and two foreign keys that link to the soldiers table and the individuals table.

RELATIONSHIPS
- soldier_id (foreign key to soldiers)
- individual_id (foreign key to individuals)
- relationship_type (varchar)

## Database Requirements

The only requirement is that it must have at least one join table, which means that it must have at least three tables, one of which will have two foreign keys (i.e. the the primary keys of the other two tables), with or without an attribute. A basic join table generally uses the naming convention of [table1]_[table2], where [table1] and [table2] are the names of the two tables that it joins. For example, if you have a table called `students` and a table called `courses`, your join table should be called `students_courses`. Nevertheless, for an attributed join table, you can just give it a name that describes the relationship that it models. For example, if you have a table called `soldiers` and a table called `individuals`, and you want to model the relationships between soldiers and individuals, you could call your join table `relationships` and have an attribute called `relationship_type` that describes the type of relationship (e.g. mother, father, spouse, sibling, friend, etc.).

## Step 2: Create Tables

Create your tables using the `CREATE TABLE` SQL command. For instance, here is what a statement would look like for the example of the soldier letters -- you can use this example as a template for your own tables. We will include the SQL query in this Markdown cell, and then provide the Python code to execute it in the next cell.

```sql
CREATE TABLE soldiers (
    id SERIAL PRIMARY KEY,
    name VARCHAR(255),
    rank VARCHAR(255),
    regiment VARCHAR(255)
);
CREATE TABLE places (
    id SERIAL PRIMARY KEY,
    name VARCHAR(255),
    location VARCHAR(255),
    latitude FLOAT,
    longitude FLOAT,
    location_description VARCHAR(255)
);
CREATE TABLE letters (
    id SERIAL PRIMARY KEY,
    date_written DATE,
    sender_id INTEGER REFERENCES soldiers(id),
    recipient_id INTEGER REFERENCES soldiers(id),
    content TEXT,
    sent_from INTEGER REFERENCES places(id),
    sent_to INTEGER REFERENCES places(id)
);
CREATE TABLE individuals (
    id SERIAL PRIMARY KEY,
    name VARCHAR(255),
    notes VARCHAR(255)
);
CREATE TABLE relationships (
    id SERIAL PRIMARY KEY,
    soldier_id INTEGER REFERENCES soldiers(id),
    individual_id INTEGER REFERENCES individuals(id),
    relationship_type VARCHAR(255)
);
CREATE TABLE soldier_places (
    soldier_id INTEGER REFERENCES soldiers(id),
    place_id INTEGER REFERENCES places(id),
    PRIMARY KEY (soldier_id, place_id)
);
```

````python
# Impliment the SQL query above in Python using psycopg2
create_table_query = """
CREATE TABLE soldiers (
    id SERIAL PRIMARY KEY,
    name VARCHAR(255),
    rank VARCHAR(255),
    regiment VARCHAR(255)
);
CREATE TABLE places (
    id SERIAL PRIMARY KEY,
    name VARCHAR(255),
    location VARCHAR(255),
    latitude FLOAT,
    longitude FLOAT,
    location_description VARCHAR(255)
);
CREATE TABLE letters (
    id SERIAL PRIMARY KEY,
    date_written DATE,
    sender_id INTEGER REFERENCES soldiers(id),
    recipient_id INTEGER REFERENCES soldiers(id),
    content TEXT,
    sent_from INTEGER REFERENCES places(id),
    sent_to INTEGER REFERENCES places(id)
);
CREATE TABLE individuals (
    id SERIAL PRIMARY KEY,
    name VARCHAR(255),
    notes VARCHAR(255)
);
CREATE TABLE relationships (
    id SERIAL PRIMARY KEY,
    soldier_id INTEGER REFERENCES soldiers(id),
    individual_id INTEGER REFERENCES individuals(id),
    relationship_type VARCHAR(255)
);
CREATE TABLE soldier_places (
    soldier_id INTEGER REFERENCES soldiers(id),
    place_id INTEGER REFERENCES places(id),
    PRIMARY KEY (soldier_id, place_id)
);
"""
cursor = conn.cursor()
cursor.execute(create_table_query)
conn.commit()

# Test the table creation by listing the tables in the database
list_tables_query = """
SELECT table_name
FROM information_schema.tables
WHERE table_schema = 'public';
"""
cursor.execute(list_tables_query)
tables = cursor.fetchall()
print(tables)
````

````python
# Create a script to load the data from the CSV files into the database tables
import csv
def load_csv_to_db(table_name, csv_file):
    with open(csv_file, 'r') as f:
        reader = csv.reader(f)
        next(reader)  # Skip the header row
        for row in reader:
            placeholders = ', '.join(['%s'] * len(row))
            insert_query = f"INSERT INTO {table_name} VALUES ({placeholders})"
            cur.execute(insert_query, row)
    conn.commit()
load_csv_to_db('soldiers', 'raw-data/soldiers.csv')
load_csv_to_db('places', 'raw-data/places.csv') 
load_csv_to_db('letters', 'raw-data/letters.csv')
load_csv_to_db('individuals', 'raw-data/individuals.csv')
load_csv_to_db('relationships', 'raw-data/relationships.csv')
load_csv_to_db('soldier_places', 'raw-data/soldier_places.csv')
````

````python
# fetch tables to verify that the data was loaded correctly
cur.execute("SELECT * FROM soldiers;")
soldiers = cur.fetchall()
print(soldiers)
````

````python
# fetch and join three tables to verify that the relationships are working correctly
join_query = """
SELECT s.name AS soldier_name, p.name AS place_name, l.content AS letter_content
FROM soldiers s
JOIN soldier_places sp ON s.id = sp.soldier_id
JOIN places p ON sp.place_id = p.id
JOIN letters l ON s.id = l.sender_id OR s.id = l.recipient_id
LIMIT 5;
"""
cur.execute(join_query)
results = cur.fetchall()
# print the results of the join query in a readable format to verify join
for row in results:
    print(f"Soldier: {row[0]}, Place: {row[1]}, Letter Content: {row[2]}")
````
